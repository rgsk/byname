# byname in a notebook

`%load_ext byname` makes every cell after it `.pyn`.

```python
%load_ext byname
```

## Records

Return locals by name, pass them by name, destructure them by name.

```python
def make(*, name: str, age: int):
    greeting = f"hi {name}"
    return (name=, age=, greeting=)


name, age = "Rahul", 26
res = make(name=, age=)
res
```

Records pass between cells. Destructure in any order, with renames:

```python
(greeting=, age=years) = res
greeting, years
```

```python
res._replace(age=27), res._asdict()
```

Equal by name and value, whichever cell built them:

```python
res == (greeting="hi Rahul", age=26, name="Rahul")
```

## Loops and comprehensions

```python
people = [make(name="a", age=30), make(name="b", age=19)]
[name for (name=, age=) in people if age > 20]
```

## Spreads

```python
extra = (city="Pune")
(**res, **extra, age=27)
```

## Magics

Magics and `!shell` lines work as in Python cells, on byname code too:

```python
%time stats = (n=len(people), mean_age=sum(p.age for p in people) / len(people))
stats
```

## Errors

Tracebacks point at the cell's own lines, quoted as written:

```python
p = (x=1, y=2)
q = p._replace(x=0)
p.x / q.x
```
